```
def sample_mean_calculator(input_dist, sample_size, n_samples):
    sample_means = []
    for i in range(n_samples):
        sample = np.random.choice(input_dist, size=sample_size, replace=False)
        sample_means.append(sample.mean())
    return sample_means
```

```
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt
import seaborn as sns

# Generate a skewed normal distribution
s = st.skewnorm.rvs(10, size=10000) # 'a=10' for a right-skewed distribution

# Plot the distribution in grey
g = sns.displot(s, color='grey', kde=True)
ax = g.ax

# Calculate mean and standard deviation
s_mean = np.mean(s)
s_std = np.std(s)

# Add mean line in blue
ax.axvline(s_mean, color='blue', linestyle='--', label=f'Mean ({s_mean:.2f})')

# Add 1.5 standard deviation lines in green
ax.axvline(s_mean + 1.5 * s_std, color='green', linestyle='-', label=f'+1.5 Std Dev ({s_mean + 1.5 * s_std:.2f})')
ax.axvline(s_mean - 1.5 * s_std, color='green', linestyle='-', label=f'-1.5 Std Dev ({s_mean - 1.5 * s_std:.2f})')

# Add 2 standard deviation lines in green
ax.axvline(s_mean + 2 * s_std, color='green', linestyle='--', label=f'+2 Std Dev ({s_mean + 2 * s_std:.2f})')
ax.axvline(s_mean - 2 * s_std, color='green', linestyle='--', label=f'-2 Std Dev ({s_mean - 2 * s_std:.2f})')

ax.set_title('Right Skewed Normal Distribution with Mean and Standard Deviations')
ax.set_xlabel('Value')
ax.set_ylabel('Density')
ax.legend()
plt.show()
```

```
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Define distribution parameters
num_students = 5000
mu_dist = 70  # Mean of the distribution
sigma_dist = 3 # Standard deviation of the distribution
x_i_value = 77 # Individual score

# 2. Generate a normal distribution
student_scores = np.random.normal(loc=mu_dist, scale=sigma_dist, size=num_students)

# 3. Calculate the Z-score for x_i_value
z_score = (x_i_value - mu_dist) / sigma_dist

print(f"The Z-score for a student with score {x_i_value} is: {z_score:.2f}")

# 4. Plot the distribution
plt.figure(figsize=(10, 6))
sns.histplot(student_scores, kde=True, color='skyblue', bins=30)

# Add lines for mean and standard deviations
plt.axvline(mu_dist, color='blue', linestyle='--', label=f'Mean ({mu_dist})')
for i in range(1, 4):
    plt.axvline(mu_dist + i * sigma_dist, color='green', linestyle=':', label=f'+{i} Std Dev ({mu_dist + i * sigma_dist:.2f})' if i==1 else "")
    plt.axvline(mu_dist - i * sigma_dist, color='red', linestyle=':', label=f'-{i} Std Dev ({mu_dist - i * sigma_dist:.2f})' if i==1 else "")

# Add line for x_i_value
plt.axvline(x_i_value, color='purple', linestyle='-', linewidth=2, label=f'x_i ({x_i_value})')

plt.title(f'Normal Distribution of Student Scores (mu={mu_dist}, sigma={sigma_dist})')
plt.xlabel('Score')
plt.ylabel('Frequency / Density')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```
import matplotlib.pyplot as plt
import seaborn as sns

# Reuse the student_scores data and parameters from the previous cell
# student_scores, mu_dist, sigma_dist, x_i_value are already defined in the kernel state

g = sns.displot(student_scores, color='gray', kde=True, bins=30)
ax = g.ax # Get the axes object from the FacetGrid

# Use the actual mean and std dev from the generated student_scores for accuracy
student_scores_mean = student_scores.mean()
student_scores_std = student_scores.std()

# Set x-axis limits dynamically based on the distribution, or use a reasonable range
min_score = student_scores_mean - 4 * student_scores_std
max_score = student_scores_mean + 4 * student_scores_std
ax.set_xlim(min_score, max_score)

plt.axvline(student_scores_mean, color='orange', label=f'Mean ({student_scores_mean:.2f})')

# Plot standard deviation lines using the actual calculated mean and std dev
for v in [-3, -2, -1, 1, 2, 3]:
    # Ensure labels are only added once to avoid legend clutter
    label_text = None
    if v == 1:
        label_text = f'+1 Std Dev ({student_scores_mean + student_scores_std:.2f})'
    elif v == -1:
        label_text = f'-1 Std Dev ({student_scores_mean - student_scores_std:.2f})'

    plt.axvline(student_scores_mean + v * student_scores_std, color='olivedrab', linestyle='--', label=label_text)


# Plot specific x_i values
# The original x_i_value from the previous turn was 77
plt.axvline(x_i_value, color='purple', label=f'x_i (previous) = {x_i_value}')
# Add the values from the user's example snippet as additional lines
plt.axvline(75, color='red', linestyle=':', label='x_i (example) = 75')
plt.axvline(85, color='blue', linestyle=':', label='x_i (example) = 85')


plt.title(f'Distribution of Student Scores (mean={student_scores_mean:.2f}, std={student_scores_std:.2f})')
plt.xlabel('Score')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```
import numpy as np
from scipy import stats

# Dataset representing scores
data = np.array([85, 90, 78, 92, 88, 76, 95, 89, 82, 80])

# --- T-test Calculation ---
# Null Hypothesis: The mean of the data is equal to 85
null_mean = 85
t_statistic, p_value = stats.ttest_1samp(data, null_mean)

print(f"--- T-test Analysis ---")
print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

# --- Decision Making ---
alpha = 0.05
if p_value < alpha:
    print("Decision: Reject the Null Hypothesis (Significant difference found).")
else:
    print("Decision: Fail to Reject the Null Hypothesis (No significant difference found).")
```
